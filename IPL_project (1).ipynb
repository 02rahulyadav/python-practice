### Now, Here comes some Basic Questions

* Which team won the most matches?

```python
team_wins=df["match_winner"].value_counts()
print(team_wins)
```

* Toss Decision Trends

```python
toss_trend=df["toss_decision"].value_counts()
print(toss_trend)
```

* Toss Winner vs Match Winner

```python
df["toss_match_winner"] = (
    df["toss_winner"] == df["match_winner"]
)

print(df["toss_match_winner"].value_counts())
```

* How do teams win? (Runs vs Wickets)

```python
win_type = df["won_by"].value_counts()

print(win_type)
```

# Key Player Performances

* 1 Most "Player of the Match" Awards

```python
pom_counts = df["player_of_the_match"].value_counts()

print(pom_counts)
```

* Top Scorers

```python
top_scorers = df["top_scorer"].value_counts()

print(top_scorers.head(10))
```

* 10 Best Bowling Figures

```python
bowling_split = df["best_bowling_figure"].str.split("--", expand=True)

df["bowling_wickets"] = bowling_split[0].astype(int)
df["bowling_runs"] = bowling_split[1].astype(int)
```

```python
best_bowling = df.sort_values(
    by=["bowling_wickets", "bowling_runs"],
    ascending=[False, True]
)

print(
    best_bowling[
        ["best_bowling", "best_bowling_figure"]
    ].head(10)
)
```

# Venue Analysis

* Most Matches Played by Venue

# Custom Questions & Insights

* Q1: Who won the highest margin by runs?

```python
runs_wins = df[df["won_by"] == "Runs"]

highest_run_margin = runs_wins.loc[
    runs_wins["margin"].idxmax()
]

print(highest_run_margin[
    ["match_id", "match_winner", "margin", "team1", "team2"]
])
```

* Q2: Which player had the highest individual score?

```python
highest_score = df.loc[df["highscore"].idxmax()]

print(highest_score[
    ["top_scorer", "highscore", "match_id"]
])
```

* Q3: Which bowler had the best bowling figures?

```python
best_figure = df.sort_values(
    by=["bowling_wickets", "bowling_runs"],
    ascending=[False, True]
).iloc[0]

print(best_figure[
    ["best_bowling", "best_bowling_figure"]
])
```

# Visualization Section 📊

Create at least 5 visualizations.


* Toss Decision Distribution

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="toss_decision")

plt.title("Toss Decision Distribution")
plt.xlabel("Toss Decision")
plt.ylabel("Count")

plt.show()
```

* Toss Winner Vs Match Winner

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="toss_match_winner")

plt.title("Toss Winner vs Match Winner")
plt.xlabel("Toss Winner Also Won Match?")
plt.ylabel("Count")

plt.show()
```

* Runs vs Wickets wins

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="won_by")

plt.title("Matches Won by Runs vs Wickets")
plt.xlabel("Won By")
plt.ylabel("Number of Matches")

plt.show()
```

* Venue-wise match count

```python

plt.figure(figsize=(12, 6))

df["venue"].value_counts().plot(kind="bar")

plt.title("Matches Played at Each Venue")
plt.xlabel("Venue")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45, ha="right")

plt.show()
```

* Matches won by each team

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    y="match_winner",
    order=df["match_winner"].value_counts().index
)

plt.title("Matches Won by Each Team")
plt.xlabel("Number of Wins")
plt.ylabel("Team")

plt.show()
```
